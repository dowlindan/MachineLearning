```python
#TODO:  Create the datasets mentioned above.
import numpy as np
Xtrain = np.array(
    [[1,1,1],
    [-2,0,4],
    [-1,1,1],
    [-1,-1,-1],
    [0,1,5],
    [0,2,3]]
)

Ytrain = np.array(
    [[1],
    [0],
    [2],
    [1],
    [0],
    [2]]
)

Xtest = np.array(
    [[0,0,0],
    [0,2,5]]
)

Ytest = np.array(
    [[1],
    [0]]
)

print(Xtrain)
print(Ytrain)
print(Xtest)
print(Ytest)
         
```

### Train a Gaussian Naive Bayes Classifier
Let's train and test a *Gaussian Naive Bayes Classifier* using the aforementioned dataset.

First use the training data to get:
1.  Compute class priors
2.  Parameterize Gaussians for each feature of each class

```python
#TODO  Compute class priors and parameterize Gaussians from training data
print("Class Priors")
classes, counts = np.unique(Ytrain, return_counts=True)
priors = counts / len(Ytrain)
print(classes, ":", counts)

class_0_mean = np.mean(np.append(Xtrain[1], Xtrain[4]))
class_1_mean = np.mean(np.append(Xtrain[0], Xtrain[3]))
class_2_mean = np.mean(np.append(Xtrain[2], Xtrain[5]))

class_0_stddev = np.std(np.append(Xtrain[1], Xtrain[4]), ddof=1)
class_1_stddev = np.std(np.append(Xtrain[0], Xtrain[3]), ddof=1)
class_2_stddev = np.std(np.append(Xtrain[2], Xtrain[5]), ddof=1)

# stddev = np.std(Xtrain, axis=0, ddof=1)
# print(mean)
# print(stddev)

PDF_0 = (1 / (class_0_stddev + np.sqrt(2*np.pi))) \
    * np.exp(-np.power((Xtest - class_0_mean),2)/(2*np.power(class_0_stddev,2)))

PDF_1 = (1 / (class_1_stddev + np.sqrt(2*np.pi))) \
    * np.exp(-np.power((Xtest - class_1_mean),2)/(2*np.power(class_1_stddev,2)))

PDF_2 = (1 / (class_2_stddev + np.sqrt(2*np.pi))) \
    * np.exp(-np.power((Xtest - class_2_mean),2)/(2*np.power(class_2_stddev,2)))

print("PDF 0")
print(PDF_0)

print("PDF 1")
print(PDF_1)

print("PDF 2")
print(PDF_2)

```
