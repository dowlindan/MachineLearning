### Nearest Neighbors
Classify testing samples using *k-nearest neighbors* from the training data (target predicting NSP, discard CLASS) . 

Your process should be:
1. Read in the data as specified in the previous part.
2. Shuffle the observations
3. Select the first 2/3 (round up) of the data for training and the remaining for testing.
4. Pre-process the observable data, as appropriate.
5. Use the training data to classify each testing sample. At the top of the next cell is a hyperparameter $K$ that specifies how many nearest neighbors to use. Feel free to play with this as well as choice of similarity or distance function.

#### Implementation Details
1. Seed the random number generator for reproducability.

```python
K = 9 #PLAY WITH THIS
#TODO
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

def knn_predict_batch(X_train, y_train, X_test):
    y_predictions = []
    for row in X_test:
        distance_neighbor_tuples = (knn_predict_single(X_train, y_train, row))
        k_nearest_neighbors = distance_neighbor_tuples[:K]
        #print(k_nearest_neighbors)
        nearest_classes = [t[1] for t in k_nearest_neighbors]
        most_common = max(set(nearest_classes), key=nearest_classes.count)
        #print(most_common)
        prediction = most_common
        y_predictions.append(prediction)
    return y_predictions
    
# For each X_test, calculate distance 
def knn_predict_single(X_train, y_train, X_row):
    distance_neighbor_tuples = []
    for i in range(len(X_train)):
        X_train_row = X_train[i]
        distance = np.linalg.norm(X_train_row - X_row)
        distance_neighbor_tuples.append((distance, y_train[i]))
    distance_neighbor_tuples.sort()
    return distance_neighbor_tuples
    
csvFilePath = './CTG.csv'

df = pd.read_csv(csvFilePath)

df = df.drop(df.index[0])
df = df.drop(columns='CLASS')

feature_cols = ['LB','AC','FM','UC','ASTV','MSTV','ALTV','MLTV','DL','DS','DP','Width','Min','Max','Nmax','Nzeros','Mode','Mean','Median',
                'Variance','Tendency']
X = df[feature_cols].to_numpy()
Y = df['NSP'].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, Y, train_size=2/3, random_state=0)

X_train_mean = np.mean(X_train, axis=0)
X_train_std = np.std(X_train, axis=0)
X_train_standardized = (X_train - X_train_mean) / X_train_std

# Predict training and testing sample using trained model
X_test_standardized = (X_test - X_train_mean) / X_train_std

y_pred = knn_predict_batch(X_train, y_train, X_test)
y_true = y_test
#print(y_pred)
```

### Evaluation
Provide the following information

- Accuracy

```python
#TODO:  Displays information requested above.
classes, counts = np.unique(y_pred, return_counts=True)
priors = counts / len(y_pred)

accuracy = np.mean(y_true == y_pred)

print("Testing Data")
print("Priors:", classes, priors)
print("Accuracy:", accuracy)


```
