### Our Data
Create a dataset with $4$ points along the line $y=4x+2$ and then one additional point not on that line.   Thus you should have five total observations, your data matrix should be $X\in \mathbb{R}^{5\times1}, Y\in \mathbb{R}^{5\times1}$

```python
#TODO
#Create your observable data matrix as X
import numpy as np

X = np.array([[0],[1],[2],[3],[4]])
Y = np.array([[2],[6],[10],[14],[15]])

print(X)
print(Y)
```

### Closed Form Linear Regression
Create simple linear regression models using the dataset mentioned in the Datasets section.  Here's some additional implementation details:
1. Don’t forget to add a bias feature!
2. **IMPORTANT** If you notice there’s issues in computing the inverse of $X^TX$ due to sparcity,

```python
#TODO Create linear regression model
X2 = np.concatenate((X, np.ones((X.shape[0],1))),axis=1)
W = (np.linalg.pinv(X2.T@X2))@(X2.T@Y)

print("X Before:")
print(X)
print("X After:")
print(X2)
print("inv(X2.T@X2)")
print((np.linalg.pinv(X2.T@X2)))
print("X2.T@Y")
print(X2.T)
print(Y)
print(X2.T@Y)

print("W")
print(W)
```

### Evaluation
Finally, using your model to make predictions for your five observations, then compute the RMSE and SMAPE for this training data.

```python
##TODO:  Make predictions Yhat
slopeGuess = W[0][0]
yIntGuess = W[1][0]

print("Slope Prediction:", slopeGuess)
print("Y Intercept Guess:", yIntGuess)

print("Y Actual:")
print(Y)

Yhat = slopeGuess*X + yIntGuess 
print("Y Prediction:")
print(Yhat)

##TODO:  Compute RMSE and SMAPE
MSE = np.mean((Y-Yhat)**2)
RMSE = MSE**(1/2)
print("RMSE:")
print(RMSE)

N = Yhat.shape[0]
SMAPE = np.mean((np.abs(Y - Yhat) / (np.abs(Y) + (np.abs(Yhat)))))
print("SMAPE:")
print(SMAPE)
```
